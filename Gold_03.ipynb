```python
from pyspark.sql import functions as F
s = spark.table("workspace.migration.silver_migration")

(s.select("sacc_code", "country").distinct()
  .write.mode("overwrite").saveAsTable("workspace.migration.dim_country"))

(s.select("geography").distinct()
  .write.mode("overwrite").saveAsTable("workspace.migration.dim_geography"))

(s.select("financial_year", "year_start").distinct()
  .write.mode("overwrite").saveAsTable("workspace.migration.dim_year"))
```

```python
(s.select("sacc_code", "geography", "financial_year", "measure", "value",
          "is_preliminary", "is_national_total")
  .write.mode("overwrite").saveAsTable("workspace.migration.fact_migration"))
```
