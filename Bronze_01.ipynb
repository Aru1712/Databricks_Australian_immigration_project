```python
from pyspark.sql import functions as F

df = (spark.read
      .option("header", True)
      .option("inferSchema", True)
      .csv("/Volumes/workspace/migration/databricks_australian_immigration_project/arrivals_tidy.csv"))

df_bronze = (df
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.lit("arrivals_tidy.csv"))
    .withColumn("_measure", F.lit("arrivals")))

df_bronze.write.mode("overwrite").saveAsTable("workspace.migration.bronze_arrivals")
```

```python
print(spark.table("workspace.migration.bronze_arrivals").count())
```
